# Hands-On Exercise 1: Hospital Patient Data Validator
### Industry: Healthcare

```python

# STEP 1: Prompt the user for all four required inputs
patient_name = input("Enter patient's name: ")
patient_id = input("Enter patient's ID: ")
age_input = input("Enter patient's age: ")
weight_input = input("Enter patient's weight: ")

# STEP 2: Use a main try-except block to handle potential errors gracefully
try:
    # --- Convert and Validate Age ---
    # Convert age input to an integer
    age = int(age_input)

    # Check if age is negative e.g -5
    if age < 0:
        raise ValueError("Age cannot be negative.")

    # --- Convert Weight ---
    # Convert weight input to a float (Handles Situation 3: text like "unknown")
    weight = float(weight_input)

    # --- Display Summary ---
    # STEP 3: Print the clean patient record if all conversions pass (Situation 4)
    print("\n========== PATIENT RECORD ==========")
    print(f"Patient Name: {patient_name}")
    print(f"Patient ID: {patient_id}")
    print(f"Age: {age}")
    print(f"Weight: {weight} kg")
    print("Status: Validated")
    print("=====================================")

# STEP 4: Catch conversion errors or raised exceptions to prevent the program from crashing
except ValueError as e:
    print(f"Error: Invalid input provided. ({e})")

```

#### Bonus Challenge

```python
# STEP 1: Main function to handle patient validation
def validate_patient_record():
    try:
        # STEP 2: Collect user inputs and remove extra leading/trailing whitespace
        name = input("Enter Patient Name: ").strip()
        patient_id = input("Enter Patient ID: ").strip()
        age_input = input("Enter Age: ").strip()
        weight_input = input("Enter Weight: ").strip()


# BONUS 1 & 2: Reject empty Patient Name or Patient ID
        if not name:
            raise ValueError("Patient Name cannot be left blank.")

        if not patient_id:
            raise ValueError("Patient ID cannot be left blank.")


# BONUS 4a: Independent exception message for Age
        try:
            # Handles Situation 1: non-numeric age input e.g 'twenty'
            age = int(age_input)
        except ValueError:
            raise ValueError("Age Error: Age must be a whole number.")

        # Handles Situation 2: negative age e.g -5
        if age < 0:
            raise ValueError("Age Error: Age cannot be negative.")


        # BONUS 3 & 4b: Reject weight <= 0 with independent Weight message
        try:
            # Handles Situation 3: non-numeric weight input e.g. 'unknown'
            weight = float(weight_input)
        except ValueError:
            raise ValueError("Weight Error: Weight must be a valid number.")

        # Bonus 3: Rejects weight that is zero or negative e.g 0 or -10
        if weight <= 0:
            raise ValueError("Weight Error: Weight must be greater than zero.")

# STEP 3: Display clean Patient Summary 
        print("=========== PATIENT RECORD ===========")
        print(f"Patient Name: {name}")
        print(f"Patient ID: {patient_id}")
        print(f"Age: {age}")
        print(f"Weight: {weight} kg")
        print("Status: Validated")
        print("=====================================")

    # STEP 4: Outer error handler to display clear, friendly messages without crashing
    except ValueError as error:
        print(f"[Validation Failed]: {error}")


# STEP 5: Run the function
validate_patient_record()

```

# Hands-On Exercise 2: E-Commerce Order Processing System

# Hands-On Exercise 3: Banking Transaction Validator
### Industry: Banking & Finance

```python
# STEP 1: Define the starting account balance
balance = 150000

# STEP 2: Wrap the transaction logic inside a try block to handle runtime input errors
try:
    # Get the withdrawal amount from the customer as a string input
    withdrawal_input = input("Enter withdrawal amount: ")

    # STEP 3: Convert the user input to a float (Handles Situation 1: non-numeric inputs like 'abc')
    withdrawal = float(withdrawal_input)

    # STEP 4: Check if withdrawal amount is zero or negative (Handles Situation 2 & 3)
    if withdrawal == 0:
        raise ValueError("Withdrawal amount must be greater than zero.")
    elif withdrawal < 0:
        raise ValueError("Withdrawal amount cannot be negative.")

    # STEP 5: Check if the customer has enough funds (Handles Situation 4: Insufficient Funds)
    if withdrawal > balance:
        raise ValueError("Insufficient funds. Withdrawal amount exceeds account balance.")

    # STEP 6: Process the transaction if all checks pass (Handles Situation 5: Successful Transaction)
    remaining_balance = balance - withdrawal

# STEP 7: Display the transaction summary
    print("=========== TRANSACTION SUMMARY ============")
    print(f"Account Balance: ₦{balance:,.2f}")
    print(f"Withdrawal: ₦{withdrawal:,.2f}")
    print("Transaction Status: Successful")
    print(f"Remaining Balance: ₦{remaining_balance:,.2f}")
    print("===========================================")

# STEP 8: Catch conversion errors and custom raised validation errors
except ValueError as error:
    # Display clear failure messages without crashing the program
    print("Transaction Failed")
    print(f"Reason: {error}")
```

#### Bonus Challenge

```python
# STEP 1: Define the starting account balance and the transaction fee
balance = 150000
transaction_fee = 100
# STEP 2: Wrap the transaction logic inside a try block to handle runtime input errors
try:
    # Get the withdrawal amount from the customer as a string input
    withdrawal_input = input("Enter withdrawal amount: ")

    # STEP 3: Convert the user input to a float (Handles Situation 1: non-numeric inputs like 'abc')
    withdrawal = float(withdrawal_input)

    # STEP 4: Check if withdrawal amount is zero or negative (Handles Situation 2 & 3)
    if withdrawal == 0:
        raise ValueError("Withdrawal amount must be greater than zero.")
    elif withdrawal < 0:
        raise ValueError("Withdrawal amount cannot be negative.")

    # STEP 5: Calculate total deduction (Withdrawal + Transaction Fee)
    total_deduction = withdrawal + transaction_fee

    # STEP 6: Validate if account balance covers BOTH withdrawal and transaction fee
    if total_deduction > balance:
        raise ValueError("Insufficient funds to cover withdrawal and transaction fee.")

    # STEP 7: Calculate final remaining balance
    remaining_balance = balance - total_deduction

    # STEP 8: Display full transaction summary including the fee
    print("=========== TRANSACTION SUMMARY ==========")
    print(f"Starting Balance: ₦{balance:,}")
    print(f"Withdrawal: ₦{withdrawal:,}")
    print(f"Transaction Fee: ₦{transaction_fee:,}")
    print("Transaction Status: Successful")
    print(f"Remaining Balance: ₦{remaining_balance:,}")
    print("==========================================")

# STEP 9: Handle all validation and conversion errors gracefully
except ValueError as error:
    print("Transaction Failed")
    print(f"Reason: {error}")
```
